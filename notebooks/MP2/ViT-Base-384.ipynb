```python
# ============================================================
# ViT-Base-patch16-384 — Fake News Image Classification
# Dataset : Clean Multimodal IFND
# Platform : Kaggle (GPU P100)
# ============================================================

# ─────────────────────────────────────────
# CELL 1 — Verify GPU & System
# ─────────────────────────────────────────
import torch, os
print("PyTorch  :", torch.__version__)
print("CUDA     :", torch.cuda.is_available())
print("GPU      :", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "CPU")
print("VRAM (GB):", round(torch.cuda.get_device_properties(0).total_memory/1e9, 2) if torch.cuda.is_available() else "—")


# ─────────────────────────────────────────
# CELL 2 — Install timm
# ─────────────────────────────────────────
import subprocess
subprocess.run(["pip", "install", "--quiet", "timm"], check=True)
print("✅ timm installed")


# ─────────────────────────────────────────
# CELL 3 — Imports & Global Config
# ─────────────────────────────────────────
import os, random, warnings
import numpy as np
import pandas as pd
from PIL import Image
from tqdm import tqdm
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torch.amp import GradScaler, autocast

import torchvision.transforms as T
import timm

from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    accuracy_score, f1_score, roc_auc_score,
    classification_report, confusion_matrix, roc_curve, auc
)

warnings.filterwarnings("ignore")

# ── Reproducibility ──────────────────────
SEED = 42
def seed_everything(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark     = False  # must be False with deterministic=True

seed_everything(SEED)

# ── Device ───────────────────────────────
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# ── Paths ────────────────────────────────
BASE       = "/kaggle/input/datasets/madhavgupta283/ifnd-multimodal-dataset"
IMG_DIR    = f"{BASE}/final_images/final_images"
TRAIN_PATH = f"{BASE}/multimodel_IFND_train.csv"
VAL_PATH   = f"{BASE}/multimodel_IFND_val.csv"
TEST_PATH  = f"{BASE}/multimodel_IFND_test.csv"
SAVE_DIR   = "/kaggle/working/models_vit"
os.makedirs(SAVE_DIR, exist_ok=True)

# ── Label mapping ────────────────────────
LABEL2ID = {"FALSE": 0, "TRUE": 1}

# ── Hyperparameters ──────────────────────
IMG_SIZE         = 384          # ViT-Base-patch16-384 native resolution
BATCH_SIZE       = 16           # safe for 384×384 on P100 16GB
EPOCHS           = 25
LR               = 1e-4         # ViT needs lower LR than CNN
LR_BACKBONE      = 1e-5         # backbone LR after unfreeze
UNFREEZE_EPOCH_1 = 3            # unfreeze top 3 transformer blocks
UNFREEZE_EPOCH_2 = 6            # unfreeze top 6 transformer blocks
PATIENCE         = 5
GRAD_CLIP        = 1.0
LABEL_SMOOTHING  = 0.05
NUM_WORKERS      = 2
MIXUP_ALPHA      = 0.2
TTA_STEPS        = 5


# ─────────────────────────────────────────
# CELL 4 — Load CSVs & Class Weights
# ─────────────────────────────────────────
train_df = pd.read_csv(TRAIN_PATH)
val_df   = pd.read_csv(VAL_PATH)
test_df  = pd.read_csv(TEST_PATH)

# Normalise labels
for d in [train_df, val_df, test_df]:
    d["Label"] = d["Label"].astype(str).str.upper().map(LABEL2ID)

# Safety check — catch any unmapped labels immediately
for name, d in [("Train", train_df), ("Val", val_df), ("Test", test_df)]:
    nulls = d["Label"].isnull().sum()
    assert nulls == 0, f"❌ {name} has {nulls} unmapped labels — check LABEL2ID mapping!"
print("✅ Label mapping verified for all splits")

# Fix image path → just the filename stem
for d in [train_df, val_df, test_df]:
    d["image_path"] = d["image_path"].apply(lambda p: os.path.basename(p))

y_train = train_df["Label"].values

# Compute class weights to handle 68/32 imbalance
cw = compute_class_weight(class_weight="balanced",
                          classes=np.array([0, 1]),
                          y=y_train)
class_weights = torch.tensor(cw, dtype=torch.float32).to(device)
print(f"Class weights  →  FALSE: {cw[0]:.4f}  |  TRUE: {cw[1]:.4f}")
print(f"Train: {len(train_df)}  |  Val: {len(val_df)}  |  Test: {len(test_df)}")


# ─────────────────────────────────────────
# CELL 5 — Transforms
# ─────────────────────────────────────────
MEAN = [0.485, 0.456, 0.406]
STD  = [0.229, 0.224, 0.225]

train_transform = T.Compose([
    T.Resize((IMG_SIZE, IMG_SIZE)),        # 380→384 (4px only)
    T.RandomHorizontalFlip(p=0.5),
    T.RandomVerticalFlip(p=0.1),
    T.ColorJitter(brightness=0.3, contrast=0.3,
                  saturation=0.2, hue=0.05),
    T.RandomRotation(degrees=15),
    T.RandomGrayscale(p=0.05),
    # domain-specific: fake news compression/sharpness artifacts
    T.GaussianBlur(kernel_size=3, sigma=(0.1, 2.0)),
    T.RandomAdjustSharpness(sharpness_factor=2, p=0.3),
    T.ToTensor(),
    T.Normalize(mean=MEAN, std=STD),
    T.RandomErasing(p=0.2, scale=(0.02, 0.15)),
])

val_test_transform = T.Compose([
    T.Resize((IMG_SIZE, IMG_SIZE)),
    T.ToTensor(),
    T.Normalize(mean=MEAN, std=STD),
])

tta_transform = T.Compose([
    T.Resize((IMG_SIZE, IMG_SIZE)),
    T.RandomHorizontalFlip(p=0.5),
    T.ColorJitter(brightness=0.1, contrast=0.1),
    T.ToTensor(),
    T.Normalize(mean=MEAN, std=STD),
])


# ─────────────────────────────────────────
# CELL 6 — Mixup Helper
# ─────────────────────────────────────────
def mixup_data(x, y, alpha=MIXUP_ALPHA):
    lam     = np.random.beta(alpha, alpha) if alpha > 0 else 1.0
    idx     = torch.randperm(x.size(0)).to(x.device)
    mixed_x = lam * x + (1 - lam) * x[idx]
    return mixed_x, y, y[idx], lam

def mixup_criterion(criterion, pred, y_a, y_b, lam):
    return lam * criterion(pred, y_a) + (1 - lam) * criterion(pred, y_b)


# ─────────────────────────────────────────
# CELL 7 — Dataset Class
# ─────────────────────────────────────────
class IFNDImageDataset(Dataset):
    def __init__(self, df, img_dir, transform):
        self.img_dir   = img_dir
        self.transform = transform
        self.records   = df[["image_path", "Label"]].reset_index(drop=True)

    def __len__(self):
        return len(self.records)

    def __getitem__(self, idx):
        row      = self.records.iloc[idx]
        img_path = os.path.join(self.img_dir, row["image_path"])
        label    = int(row["Label"])
        try:
            img = Image.open(img_path).convert("RGB")
        except Exception:
            # graceful fallback — should not happen after preprocessing
            img = Image.fromarray(np.zeros((IMG_SIZE, IMG_SIZE, 3), dtype=np.uint8))
        img = self.transform(img)
        return img, torch.tensor(label, dtype=torch.long)


train_ds = IFNDImageDataset(train_df, IMG_DIR, train_transform)
val_ds   = IFNDImageDataset(val_df,   IMG_DIR, val_test_transform)
test_ds  = IFNDImageDataset(test_df,  IMG_DIR, val_test_transform)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,
                          num_workers=NUM_WORKERS, pin_memory=True,
                          persistent_workers=True)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False,
                          num_workers=NUM_WORKERS, pin_memory=True,
                          persistent_workers=True)
test_loader  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False,
                          num_workers=NUM_WORKERS, pin_memory=True,
                          persistent_workers=True)

print(f"Batches  →  Train: {len(train_loader)}  |  Val: {len(val_loader)}  |  Test: {len(test_loader)}")


# ─────────────────────────────────────────
# CELL 8 — Model Definition
# ─────────────────────────────────────────
class ViTClassifier(nn.Module):
    """
    ViT-Base-patch16-384 with custom classification head.
    • GELU activation — matches ViT internals
    • LayerNorm in head — better than BatchNorm for transformers
    • encode() exposed — used by Fusion Image Model later
    • feature_dim = 768 — exposed for fusion
    """
    def __init__(self, num_classes=2, dropout=0.3):
        super().__init__()

        # Load pretrained ViT-Base-384 via timm (num_classes=0 → raw CLS token)
        self.backbone    = timm.create_model(
            "vit_base_patch16_384",
            pretrained=True,
            num_classes=0,
        )
        self.feature_dim = self.backbone.num_features   # 768

        self.classifier = nn.Sequential(
            nn.LayerNorm(self.feature_dim),   # ViT prefers LayerNorm
            nn.Dropout(p=dropout),
            nn.Linear(self.feature_dim, 512),
            nn.GELU(),                        # GELU matches ViT internals
            nn.Dropout(p=dropout / 2),        # 0.15
            nn.Linear(512, num_classes)
        )

    def encode(self, x):
        """Returns 768-d CLS token — used by Fusion Image Model."""
        return self.backbone(x)               # [B, 768]

    def forward(self, x):
        return self.classifier(self.encode(x))


model = ViTClassifier(num_classes=2, dropout=0.3).to(device)
print(f"Feature dim : {model.feature_dim}")
print(f"Total params: {sum(p.numel() for p in model.parameters()):,}")


# ─────────────────────────────────────────
# CELL 9 — Freeze / Unfreeze Helpers
# ─────────────────────────────────────────
def freeze_backbone(model):
    """Freeze entire ViT backbone — train classifier head only."""
    for p in model.backbone.parameters():
        p.requires_grad = False

def unfreeze_top_vit_blocks(model, num_blocks):
    """
    Unfreeze last num_blocks transformer blocks of ViT-Base (12 total).
    Also always unfreeze the final norm layer.
    """
    blocks = list(model.backbone.blocks)
    for block in blocks[-num_blocks:]:
        for p in block.parameters():
            p.requires_grad = True
    # always unfreeze final norm
    for p in model.backbone.norm.parameters():
        p.requires_grad = True

def count_trainable(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

freeze_backbone(model)
print(f"Phase-1 trainable params: {count_trainable(model):,}")


# ─────────────────────────────────────────
# CELL 10 — Loss, Optimizer, Scheduler
# ─────────────────────────────────────────
criterion = nn.CrossEntropyLoss(
    weight=class_weights,
    label_smoothing=LABEL_SMOOTHING
)

def get_optimizer(model, phase=1):
    if phase == 1:
        return torch.optim.AdamW(
            filter(lambda p: p.requires_grad, model.parameters()),
            lr=LR, weight_decay=1e-4
        )
    backbone_params   = [p for p in model.backbone.parameters() if p.requires_grad]
    classifier_params = list(model.classifier.parameters())
    return torch.optim.AdamW([
        {"params": backbone_params,   "lr": LR_BACKBONE},
        {"params": classifier_params, "lr": LR},
    ], weight_decay=1e-4)

def get_scheduler(optimizer):
    """ReduceLROnPlateau — steps on val_F1."""
    return torch.optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, mode="max", factor=0.5,
        patience=2, min_lr=1e-6
    )

optimizer = get_optimizer(model, phase=1)
scaler    = GradScaler(device="cuda", init_scale=1024)  # lower init prevents NaN early
scheduler = get_scheduler(optimizer)


# ─────────────────────────────────────────
# CELL 11 — Evaluation Helper
# ─────────────────────────────────────────
@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    all_preds, all_labels, all_probs, losses = [], [], [], []
    for imgs, labels in loader:
        imgs, labels = imgs.to(device), labels.to(device)
        # ✅ pure fp32 — no AMP in eval (same as EfficientNet working code)
        logits   = model(imgs).float()
        loss     = criterion(logits, labels)
        logits_f = torch.clamp(logits, min=-50, max=50)
        probs    = F.softmax(logits_f, dim=1)[:, 1]
        losses.append(loss.item())
        all_preds.extend(torch.argmax(logits_f, dim=1).cpu().numpy())
        all_labels.extend(labels.cpu().numpy())
        all_probs.extend(probs.cpu().numpy())
    y_true = np.array(all_labels)
    y_pred = np.array(all_preds)
    y_prob = np.nan_to_num(np.array(all_probs), nan=0.5)
    return (np.mean(losses),
            f1_score(y_true, y_pred, average="macro"),
            accuracy_score(y_true, y_pred),
            roc_auc_score(y_true, y_prob),
            y_true, y_pred, y_prob)


# ─────────────────────────────────────────
# CELL 12 — Test-Time Augmentation (TTA)
# ─────────────────────────────────────────
@torch.no_grad()
def evaluate_tta(model, df, img_dir, n_steps=TTA_STEPS):
    """Average probabilities over n_steps augmented passes."""
    model.eval()
    all_probs_accum = None
    all_labels      = None

    for _ in range(n_steps):
        tta_ds     = IFNDImageDataset(df, img_dir, tta_transform)
        tta_loader = DataLoader(tta_ds, batch_size=BATCH_SIZE,
                                shuffle=False, num_workers=NUM_WORKERS)
        step_probs, step_labels = [], []
        for imgs, labels in tta_loader:
            imgs = imgs.to(device)
            with autocast(device_type="cuda"):
                logits = model(imgs)
            logits_f = torch.clamp(logits.float(), min=-50, max=50)
            step_probs.extend(
                F.softmax(logits_f, dim=1)[:, 1].cpu().numpy()
            )
            step_labels.extend(labels.numpy())

        step_probs = np.array(step_probs)
        if all_probs_accum is None:
            all_probs_accum = step_probs
            all_labels      = np.array(step_labels)
        else:
            all_probs_accum += step_probs

    avg_probs  = all_probs_accum / n_steps
    y_pred_tta = (avg_probs >= 0.5).astype(int)

    return (accuracy_score(all_labels, y_pred_tta),
            f1_score(all_labels, y_pred_tta, average="macro"),
            roc_auc_score(all_labels, avg_probs),
            all_labels, y_pred_tta, avg_probs)


# ─────────────────────────────────────────
# CELL 13 — Training Loop
# ─────────────────────────────────────────
best_val_f1   = -1.0
wait          = 0
best_path     = os.path.join(SAVE_DIR, "vit_base_384-best.pt")
best_f1_epoch = 1
train_accs, val_accs = [], []

print("=" * 60)
print("Starting ViT-Base-384 Training")
print("=" * 60)

for epoch in range(1, EPOCHS + 1):

    # ── Progressive unfreeze ─────────────
    if epoch == UNFREEZE_EPOCH_1:
        print(f"\n🔓 Epoch {epoch}: Unfreezing top-3 ViT blocks...")
        unfreeze_top_vit_blocks(model, num_blocks=3)
        optimizer = get_optimizer(model, phase=2)
        scaler    = GradScaler(device="cuda")   # reset scaler with new optimizer
        scheduler = get_scheduler(optimizer)
        print(f"   Trainable params: {count_trainable(model):,}")

    elif epoch == UNFREEZE_EPOCH_2:
        print(f"\n🔓 Epoch {epoch}: Unfreezing top-6 ViT blocks...")
        unfreeze_top_vit_blocks(model, num_blocks=6)
        optimizer = get_optimizer(model, phase=2)
        scaler    = GradScaler(device="cuda")   # reset scaler with new optimizer
        scheduler = get_scheduler(optimizer)
        print(f"   Trainable params: {count_trainable(model):,}")

    # ── Train one epoch ──────────────────
    model.train()
    running_loss = 0.0
    correct = total = 0

    pbar = tqdm(train_loader, desc=f"Epoch {epoch:02d}/{EPOCHS}")
    for imgs, labels in pbar:
        imgs, labels = imgs.to(device), labels.to(device)

        imgs, labels_a, labels_b, lam = mixup_data(imgs, labels)

        optimizer.zero_grad(set_to_none=True)

        with autocast(device_type="cuda"):
            logits = model(imgs)
            loss   = mixup_criterion(criterion, logits, labels_a, labels_b, lam)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), GRAD_CLIP)
        scaler.step(optimizer)
        scaler.update()

        running_loss += loss.item()
        with torch.no_grad():
            preds = torch.argmax(logits.float(), dim=1)
        # mixup-aware accuracy
        correct += (lam * (preds == labels_a).float() +
                    (1 - lam) * (preds == labels_b).float()).sum().item()
        total   += labels_a.size(0)

        pbar.set_postfix({
            "loss": f"{running_loss / (pbar.n or 1):.4f}",
            "acc" : f"{correct / total:.4f}"
        })

    train_acc = correct / total
    train_accs.append(train_acc)

    # ── Validation ───────────────────────
    val_loss, val_f1, val_acc, val_auc, _, _, _ = evaluate(model, val_loader)
    val_accs.append(val_acc)

    scheduler.step(val_f1)   # ReduceLROnPlateau steps on val_F1

    print(f"\nEpoch {epoch:02d} | "
          f"train_acc={train_acc:.4f} | "
          f"val_acc={val_acc:.4f} | "
          f"val_F1={val_f1:.4f} | "
          f"val_AUC={val_auc:.4f} | "
          f"val_loss={val_loss:.4f}")

    # ── Checkpoint & early stopping ──────
    if val_f1 > best_val_f1:
        best_val_f1   = val_f1
        best_f1_epoch = epoch
        wait = 0
        torch.save({
            "epoch"           : epoch,
            "model_state_dict": model.state_dict(),
            "val_f1"          : val_f1,
            "val_acc"         : val_acc,
            "feature_dim"     : model.feature_dim,
        }, best_path)
        print(f"✅ Best model saved  →  val_F1={val_f1:.4f}")
    else:
        wait += 1
        print(f"⏳ No improvement ({wait}/{PATIENCE})")
        if wait >= PATIENCE:
            print("⏹️  Early stopping triggered.")
            break

print(f"\nBest Val macro-F1: {best_val_f1:.4f}  (epoch {best_f1_epoch})")


# ─────────────────────────────────────────
# CELL 14 — Training vs Validation Accuracy Plot
# ─────────────────────────────────────────
best_epoch = best_f1_epoch

plt.figure(figsize=(9, 5))
plt.plot(range(1, len(train_accs)+1), [a*100 for a in train_accs],
         label="Training Accuracy", linewidth=2)
plt.plot(range(1, len(val_accs)+1),   [a*100 for a in val_accs],
         label="Validation Accuracy", linewidth=2, color="orange")
plt.axvline(x=best_epoch, linestyle="--", color="gray")
plt.scatter([best_epoch], [val_accs[best_epoch-1]*100],
            color="red", zorder=5, s=80,
            label=f"Best Saved (Epoch {best_epoch})\n"
                  f"Val: {val_accs[best_epoch-1]*100:.2f}%\n"
                  f"Train: {train_accs[best_epoch-1]*100:.2f}%")
plt.xlabel("Epoch"); plt.ylabel("Accuracy (%)")
plt.title("ViT-Base-384 — Training vs Validation Accuracy")
plt.legend(); plt.grid(alpha=0.3)
plt.tight_layout()
acc_plot_path = os.path.join(SAVE_DIR, "vit_train_val_accuracy.png")
plt.savefig(acc_plot_path, dpi=400, bbox_inches="tight")
plt.show()
print("Saved →", acc_plot_path)


# ─────────────────────────────────────────
# CELL 15 — Load Best & Evaluate on Test
# ─────────────────────────────────────────
checkpoint = torch.load(best_path, map_location=device, weights_only=True)
model.load_state_dict(checkpoint["model_state_dict"])
model.to(device)
print(f"Loaded best model from epoch {checkpoint['epoch']}")

# Standard single-pass
test_loss, test_f1, test_acc, test_auc, y_true, y_pred, y_prob = evaluate(model, test_loader)

print(f"\n── Standard Inference ──────────────────────────────────")
print(f"TEST  →  Accuracy: {test_acc:.4f}  |  Macro-F1: {test_f1:.4f}  |  ROC-AUC: {test_auc:.4f}")
print("\nClassification Report:\n",
      classification_report(y_true, y_pred,
                             target_names=["FALSE", "TRUE"], digits=4))
print("Confusion Matrix:\n", confusion_matrix(y_true, y_pred))

# TTA
print(f"\n── Test-Time Augmentation ({TTA_STEPS} passes) ──────────────────")
tta_acc, tta_f1, tta_auc, tta_true, tta_pred, tta_prob = evaluate_tta(
    model, test_df, IMG_DIR, n_steps=TTA_STEPS
)
print(f"TTA   →  Accuracy: {tta_acc:.4f}  |  Macro-F1: {tta_f1:.4f}  |  ROC-AUC: {tta_auc:.4f}")
print("\nTTA Classification Report:\n",
      classification_report(tta_true, tta_pred,
                             target_names=["FALSE", "TRUE"], digits=4))
print("TTA Confusion Matrix:\n", confusion_matrix(tta_true, tta_pred))


# ─────────────────────────────────────────
# CELL 16 — Train vs Test ROC Curve
# ─────────────────────────────────────────
@torch.no_grad()
def get_roc_data(loader, name="", mdl=None):
    if mdl is None: mdl = model
    mdl.eval()
    y_true_list, y_score_list = [], []
    for imgs, labels in tqdm(loader, desc=f"ROC ({name})"):
        imgs = imgs.to(device)
        with autocast(device_type="cuda"):
            logits = mdl(imgs)
        probs = F.softmax(torch.clamp(logits.float(), min=-50, max=50), dim=1)[:, 1].cpu().numpy()
        y_true_list.extend(labels.numpy())
        y_score_list.extend(probs)
    fpr, tpr, _ = roc_curve(y_true_list, y_score_list)
    return fpr, tpr, auc(fpr, tpr)

fpr_train, tpr_train, auc_train = get_roc_data(train_loader, "Train")
fpr_test,  tpr_test,  auc_test  = get_roc_data(test_loader,  "Test")

plt.figure(figsize=(7, 6))
plt.plot(fpr_train, tpr_train, label=f"Train (AUC = {auc_train:.4f})", linewidth=2.5)
plt.plot(fpr_test,  tpr_test,  label=f"Test  (AUC = {auc_test:.4f})",  linewidth=2.5)
plt.plot([0,1], [0,1], linestyle="--", color="gray")
plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
plt.title("Train vs Test ROC Curve — ViT-Base-384")
plt.legend(loc="lower right"); plt.grid(alpha=0.3)
plt.tight_layout()
roc_path = os.path.join(SAVE_DIR, "vit_train_vs_test_roc.png")
plt.savefig(roc_path, dpi=400, bbox_inches="tight")
plt.show()
print("Saved →", roc_path)

gap = abs(auc_train - auc_test)
print(f"\nTrain AUC : {auc_train:.4f}")
print(f"Test  AUC : {auc_test:.4f}")
print(f"Gap       : {gap:.4f}  →  ", end="")
if   gap <= 0.02: print("Excellent generalization ✅")
elif gap <= 0.05: print("Slight overfitting — acceptable ⚠️")
else:             print("Significant overfitting — review regularization 🚨")


# ─────────────────────────────────────────
# CELL 17 — Save Per-Sample Predictions
# ─────────────────────────────────────────
preds_df = pd.DataFrame({
    "y_true"      : y_true,
    "y_prob_TRUE" : y_prob,
    "y_pred"      : y_pred,
    "y_pred_tta"  : tta_pred,
    "y_prob_tta"  : tta_prob,
})
preds_path = os.path.join(SAVE_DIR, "vit_test_predictions.csv")
preds_df.to_csv(preds_path, index=False)
print("✅ Saved test predictions →", preds_path)


# ─────────────────────────────────────────
# CELL 18 — Final Summary
# ─────────────────────────────────────────
print("\n" + "="*60)
print("        ViT-Base-384 — Final Results Summary")
print("="*60)
print(f"  Standard Inference")
print(f"    Test Accuracy   :  {test_acc:.4f}")
print(f"    Test Macro-F1   :  {test_f1:.4f}")
print(f"    Test ROC-AUC    :  {test_auc:.4f}")
print(f"    Train ROC-AUC   :  {auc_train:.4f}")
print(f"  TTA ({TTA_STEPS} passes)")
print(f"    TTA Accuracy    :  {tta_acc:.4f}")
print(f"    TTA Macro-F1    :  {tta_f1:.4f}")
print(f"    TTA ROC-AUC     :  {tta_auc:.4f}")
print(f"  Best Val F1       :  {best_val_f1:.4f}  (epoch {best_f1_epoch})")
print(f"  Model saved at    :  {best_path}")
print("="*60)
```
