```python
# ============================================================
# Fusion Image Model — EfficientNet-B3 + ViT-Base-384
# Dataset : Clean Multimodal IFND
# Platform : Kaggle (GPU P100)
# ============================================================

# ─────────────────────────────────────────
# CELL 1 — Verify GPU & System
# ─────────────────────────────────────────
import torch, os
print("PyTorch  :", torch.__version__)
print("CUDA     :", torch.cuda.is_available())
print("GPU      :", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "CPU")
print("VRAM (GB):", round(torch.cuda.get_device_properties(0).total_memory/1e9, 2) if torch.cuda.is_available() else "—")


# ─────────────────────────────────────────
# CELL 2 — Install timm
# ─────────────────────────────────────────
import subprocess
subprocess.run(["pip", "install", "--quiet", "timm"], check=True)
print("✅ timm installed")


# ─────────────────────────────────────────
# CELL 3 — Imports & Global Config
# ─────────────────────────────────────────
import os, random, warnings
import numpy as np
import pandas as pd
from PIL import Image
from tqdm import tqdm
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
from torch.amp import GradScaler, autocast

import torchvision.transforms as T
from torchvision.models import efficientnet_b3, EfficientNet_B3_Weights
import timm

from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    accuracy_score, f1_score, roc_auc_score,
    classification_report, confusion_matrix, roc_curve, auc
)

warnings.filterwarnings("ignore")

# ── Reproducibility ──────────────────────
SEED = 42
def seed_everything(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark     = False

seed_everything(SEED)

# ── Device ───────────────────────────────
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", device)

# ── Paths ────────────────────────────────
# ⚠️ CHANGE BASE path according to your account
BASE       = "/kaggle/input/datasets/madhavgupta90/ifnd-dataset"
IMG_DIR    = f"{BASE}/final_images/final_images"
TRAIN_PATH = f"{BASE}/multimodel_IFND_train.csv"
VAL_PATH   = f"{BASE}/multimodel_IFND_val.csv"
TEST_PATH  = f"{BASE}/multimodel_IFND_test.csv"

# ⚠️ CHANGE to your uploaded .pt files path
EFFNET_CKPT = "/kaggle/input/datasets/madhavgupta90/image-models/efficientnet_b3-best.pt"
VIT_CKPT    = "/kaggle/input/datasets/madhavgupta90/image-models/vit_base_384-best.pt"

SAVE_DIR    = "/kaggle/working/models_fusion_image"
os.makedirs(SAVE_DIR, exist_ok=True)

# ── Label mapping ────────────────────────
LABEL2ID = {"FALSE": 0, "TRUE": 1}

# ── Hyperparameters ──────────────────────
IMG_SIZE_EFF = 380          # EfficientNet-B3 native
IMG_SIZE_VIT = 384          # ViT-Base-384 native
BATCH_SIZE   = 16           # safe for P100
EPOCHS       = 20
LR           = 3e-4         # Fusion head LR
PATIENCE     = 5
GRAD_CLIP    = 1.0
LABEL_SMOOTHING = 0.05
NUM_WORKERS  = 2
MIXUP_ALPHA  = 0.2
TTA_STEPS    = 5

# Fusion feature dims (from individual models)
EFFNET_DIM   = 1536
VIT_DIM      = 768
FUSION_DIM   = EFFNET_DIM + VIT_DIM  # 2304


# ─────────────────────────────────────────
# CELL 4 — Load CSVs & Class Weights
# ─────────────────────────────────────────
train_df = pd.read_csv(TRAIN_PATH)
val_df   = pd.read_csv(VAL_PATH)
test_df  = pd.read_csv(TEST_PATH)

for d in [train_df, val_df, test_df]:
    d["Label"] = d["Label"].astype(str).str.upper().map(LABEL2ID)

for name, d in [("Train", train_df), ("Val", val_df), ("Test", test_df)]:
    nulls = d["Label"].isnull().sum()
    assert nulls == 0, f"❌ {name} has {nulls} unmapped labels!"
print("✅ Label mapping verified for all splits")

for d in [train_df, val_df, test_df]:
    d["image_path"] = d["image_path"].apply(lambda p: os.path.basename(p))

y_train = train_df["Label"].values
cw = compute_class_weight(class_weight="balanced",
                          classes=np.array([0, 1]), y=y_train)
class_weights = torch.tensor(cw, dtype=torch.float32).to(device)
print(f"Class weights  →  FALSE: {cw[0]:.4f}  |  TRUE: {cw[1]:.4f}")
print(f"Train: {len(train_df)}  |  Val: {len(val_df)}  |  Test: {len(test_df)}")


# ─────────────────────────────────────────
# CELL 5 — Dual Transforms
# ─────────────────────────────────────────
MEAN = [0.485, 0.456, 0.406]
STD  = [0.229, 0.224, 0.225]

# EfficientNet transform (380×380)
eff_train_transform = T.Compose([
    T.Resize((IMG_SIZE_EFF, IMG_SIZE_EFF)),
    T.RandomHorizontalFlip(p=0.5),
    T.RandomVerticalFlip(p=0.1),
    T.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.2, hue=0.05),
    T.RandomRotation(degrees=15),
    T.RandomGrayscale(p=0.05),
    T.GaussianBlur(kernel_size=3, sigma=(0.1, 2.0)),
    T.RandomAdjustSharpness(sharpness_factor=2, p=0.3),
    T.ToTensor(),
    T.Normalize(mean=MEAN, std=STD),
    T.RandomErasing(p=0.2, scale=(0.02, 0.15)),
])
eff_val_transform = T.Compose([
    T.Resize((IMG_SIZE_EFF, IMG_SIZE_EFF)),
    T.ToTensor(),
    T.Normalize(mean=MEAN, std=STD),
])
eff_tta_transform = T.Compose([
    T.Resize((IMG_SIZE_EFF, IMG_SIZE_EFF)),
    T.RandomHorizontalFlip(p=0.5),
    T.ColorJitter(brightness=0.1, contrast=0.1),
    T.ToTensor(),
    T.Normalize(mean=MEAN, std=STD),
])

# ViT transform (384×384)
vit_train_transform = T.Compose([
    T.Resize((IMG_SIZE_VIT, IMG_SIZE_VIT)),
    T.RandomHorizontalFlip(p=0.5),
    T.RandomVerticalFlip(p=0.1),
    T.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.2, hue=0.05),
    T.RandomRotation(degrees=15),
    T.RandomGrayscale(p=0.05),
    T.GaussianBlur(kernel_size=3, sigma=(0.1, 2.0)),
    T.RandomAdjustSharpness(sharpness_factor=2, p=0.3),
    T.ToTensor(),
    T.Normalize(mean=MEAN, std=STD),
    T.RandomErasing(p=0.2, scale=(0.02, 0.15)),
])
vit_val_transform = T.Compose([
    T.Resize((IMG_SIZE_VIT, IMG_SIZE_VIT)),
    T.ToTensor(),
    T.Normalize(mean=MEAN, std=STD),
])
vit_tta_transform = T.Compose([
    T.Resize((IMG_SIZE_VIT, IMG_SIZE_VIT)),
    T.RandomHorizontalFlip(p=0.5),
    T.ColorJitter(brightness=0.1, contrast=0.1),
    T.ToTensor(),
    T.Normalize(mean=MEAN, std=STD),
])


# ─────────────────────────────────────────
# CELL 6 — Dual Dataset Class
# ─────────────────────────────────────────
class DualImageDataset(Dataset):
    """
    Returns TWO versions of each image:
    - img_eff : 380×380 for EfficientNet-B3
    - img_vit : 384×384 for ViT-Base-384
    """
    def __init__(self, df, img_dir, eff_transform, vit_transform):
        self.img_dir       = img_dir
        self.eff_transform = eff_transform
        self.vit_transform = vit_transform
        self.records       = df[["image_path", "Label"]].reset_index(drop=True)

    def __len__(self):
        return len(self.records)

    def __getitem__(self, idx):
        row      = self.records.iloc[idx]
        img_path = os.path.join(self.img_dir, row["image_path"])
        label    = int(row["Label"])
        try:
            img = Image.open(img_path).convert("RGB")
        except Exception:
            img = Image.fromarray(np.zeros((384, 384, 3), dtype=np.uint8))
        img_eff = self.eff_transform(img)
        img_vit = self.vit_transform(img)
        return img_eff, img_vit, torch.tensor(label, dtype=torch.long)


train_ds = DualImageDataset(train_df, IMG_DIR, eff_train_transform, vit_train_transform)
val_ds   = DualImageDataset(val_df,   IMG_DIR, eff_val_transform,   vit_val_transform)
test_ds  = DualImageDataset(test_df,  IMG_DIR, eff_val_transform,   vit_val_transform)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,
                          num_workers=NUM_WORKERS, pin_memory=True,
                          persistent_workers=True)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False,
                          num_workers=NUM_WORKERS, pin_memory=True,
                          persistent_workers=True)
test_loader  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False,
                          num_workers=NUM_WORKERS, pin_memory=True,
                          persistent_workers=True)
print(f"Batches  →  Train: {len(train_loader)}  |  Val: {len(val_loader)}  |  Test: {len(test_loader)}")


# ─────────────────────────────────────────
# CELL 7 — Individual Model Definitions
# ─────────────────────────────────────────
class EfficientNetB3Classifier(nn.Module):
    def __init__(self, num_classes=2, dropout=0.3):
        super().__init__()
        base             = efficientnet_b3(weights=EfficientNet_B3_Weights.IMAGENET1K_V1)
        self.backbone    = base.features
        self.pool        = base.avgpool
        self.feature_dim = base.classifier[1].in_features  # 1536
        self.classifier  = nn.Sequential(
            nn.Dropout(p=dropout),
            nn.Linear(self.feature_dim, 512),
            nn.BatchNorm1d(512),
            nn.SiLU(inplace=True),
            nn.Dropout(p=dropout / 2),
            nn.Linear(512, num_classes)
        )
    def encode(self, x):
        x = self.backbone(x)
        x = self.pool(x)
        return torch.flatten(x, 1)   # [B, 1536]
    def forward(self, x):
        return self.classifier(self.encode(x))


class ViTClassifier(nn.Module):
    def __init__(self, num_classes=2, dropout=0.3):
        super().__init__()
        self.backbone    = timm.create_model(
            "vit_base_patch16_384", pretrained=False, num_classes=0)
        self.feature_dim = self.backbone.num_features   # 768
        self.classifier  = nn.Sequential(
            nn.LayerNorm(self.feature_dim),
            nn.Dropout(p=dropout),
            nn.Linear(self.feature_dim, 512),
            nn.GELU(),
            nn.Dropout(p=dropout / 2),
            nn.Linear(512, num_classes)
        )
    def encode(self, x):
        return self.backbone(x)   # [B, 768]
    def forward(self, x):
        return self.classifier(self.encode(x))


# ─────────────────────────────────────────
# CELL 8 — Load Pretrained Weights
# ─────────────────────────────────────────
print("Loading EfficientNet-B3 weights...")
effnet_model = EfficientNetB3Classifier(num_classes=2, dropout=0.3)
eff_ckpt     = torch.load(EFFNET_CKPT, map_location="cpu", weights_only=True)
effnet_model.load_state_dict(eff_ckpt["model_state_dict"])
effnet_model = effnet_model.to(device)
effnet_model.eval()
# Freeze completely — only fusion head trains
for p in effnet_model.parameters():
    p.requires_grad = False
print(f"✅ EfficientNet-B3 loaded  (feature_dim={eff_ckpt['feature_dim']}, "
      f"val_f1={eff_ckpt['val_f1']:.4f}, epoch={eff_ckpt['epoch']})")

print("Loading ViT-Base-384 weights...")
vit_model  = ViTClassifier(num_classes=2, dropout=0.3)
vit_ckpt   = torch.load(VIT_CKPT, map_location="cpu", weights_only=True)
vit_model.load_state_dict(vit_ckpt["model_state_dict"])
vit_model  = vit_model.to(device)
vit_model.eval()
# Freeze completely — only fusion head trains
for p in vit_model.parameters():
    p.requires_grad = False
print(f"✅ ViT-Base-384 loaded     (feature_dim={vit_ckpt['feature_dim']}, "
      f"val_f1={vit_ckpt['val_f1']:.4f}, epoch={vit_ckpt['epoch']})")


# ─────────────────────────────────────────
# CELL 9 — Fusion Image Model
# ─────────────────────────────────────────
class FusionImageModel(nn.Module):
    """
    Fusion of EfficientNet-B3 (1536-d) + ViT-Base-384 (768-d)
    → Concatenate → 2304-d → Fusion Head → 2 classes

    Both backbones are FROZEN — only fusion head is trained.
    expose encode() → 2304-d features for Final Multimodal Model
    """
    def __init__(self, effnet, vit, num_classes=2, dropout=0.4):
        super().__init__()
        self.effnet      = effnet   # frozen
        self.vit         = vit      # frozen
        self.feature_dim = FUSION_DIM  # 2304

        self.fusion_head = nn.Sequential(
            nn.LayerNorm(FUSION_DIM),    # ← yeh add karo
            nn.Dropout(p=dropout),
            nn.Linear(FUSION_DIM, 1024),
            nn.GELU(),
            nn.Dropout(p=dropout / 2),
            nn.Linear(1024, num_classes)
        )

    def encode(self, img_eff, img_vit):
        """Returns 2304-d fused feature — used by Final Multimodal Model."""
        with torch.no_grad():
            feat_eff = self.effnet.encode(img_eff)   # [B, 1536]
            feat_vit = self.vit.encode(img_vit)       # [B, 768]
        return torch.cat([feat_eff, feat_vit], dim=1) # [B, 2304]

    def forward(self, img_eff, img_vit):
        fused = self.encode(img_eff, img_vit)
        return self.fusion_head(fused)


fusion_model = FusionImageModel(effnet_model, vit_model,
                                num_classes=2, dropout=0.4).to(device)

trainable = sum(p.numel() for p in fusion_model.parameters() if p.requires_grad)
total     = sum(p.numel() for p in fusion_model.parameters())
print(f"\nFusion Image Model")
print(f"  Total params     : {total:,}")
print(f"  Trainable params : {trainable:,}  (fusion head only)")
print(f"  Feature dim      : {fusion_model.feature_dim}")


# ─────────────────────────────────────────
# CELL 10 — Mixup Helper
# ─────────────────────────────────────────
def mixup_data(x1, x2, y, alpha=MIXUP_ALPHA):
    lam      = np.random.beta(alpha, alpha) if alpha > 0 else 1.0
    idx      = torch.randperm(x1.size(0)).to(x1.device)
    mixed_x1 = lam * x1 + (1 - lam) * x1[idx]
    mixed_x2 = lam * x2 + (1 - lam) * x2[idx]
    return mixed_x1, mixed_x2, y, y[idx], lam

def mixup_criterion(criterion, pred, y_a, y_b, lam):
    return lam * criterion(pred, y_a) + (1 - lam) * criterion(pred, y_b)


# ─────────────────────────────────────────
# CELL 11 — Loss, Optimizer, Scheduler
# ─────────────────────────────────────────
criterion = nn.CrossEntropyLoss(
    weight=class_weights,
    label_smoothing=LABEL_SMOOTHING
)

optimizer = torch.optim.AdamW(
    filter(lambda p: p.requires_grad, fusion_model.parameters()),
    lr=LR, weight_decay=1e-4
)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode="max", factor=0.5, patience=2, min_lr=1e-6
)
scaler = GradScaler(device="cuda", init_scale=1024)


# ─────────────────────────────────────────
# CELL 12 — Evaluation Helper
# ─────────────────────────────────────────
@torch.no_grad()
def evaluate(model, loader):
    model.eval()
    all_preds, all_labels, all_probs, losses = [], [], [], []
    for img_eff, img_vit, labels in loader:
        img_eff = img_eff.to(device)
        img_vit = img_vit.to(device)
        labels  = labels.to(device)
        # ✅ pure fp32 — no AMP in eval
        logits   = model(img_eff, img_vit).float()
        loss     = criterion(logits, labels)
        logits_f = torch.clamp(logits, min=-50, max=50)
        probs    = F.softmax(logits_f, dim=1)[:, 1]
        losses.append(loss.item())
        all_preds.extend(torch.argmax(logits_f, dim=1).cpu().numpy())
        all_labels.extend(labels.cpu().numpy())
        all_probs.extend(probs.cpu().numpy())
    y_true = np.array(all_labels)
    y_pred = np.array(all_preds)
    y_prob = np.nan_to_num(np.array(all_probs), nan=0.5)
    return (np.mean(losses),
            f1_score(y_true, y_pred, average="macro"),
            accuracy_score(y_true, y_pred),
            roc_auc_score(y_true, y_prob),
            y_true, y_pred, y_prob)


# ─────────────────────────────────────────
# CELL 13 — TTA Helper
# ─────────────────────────────────────────
@torch.no_grad()
def evaluate_tta(model, df, img_dir, n_steps=TTA_STEPS):
    model.eval()
    all_probs_accum = None
    all_labels      = None

    for _ in range(n_steps):
        tta_ds     = DualImageDataset(df, img_dir, eff_tta_transform, vit_tta_transform)
        tta_loader = DataLoader(tta_ds, batch_size=BATCH_SIZE,
                                shuffle=False, num_workers=NUM_WORKERS)
        step_probs, step_labels = [], []
        for img_eff, img_vit, labels in tta_loader:
            img_eff = img_eff.to(device)
            img_vit = img_vit.to(device)
            with autocast(device_type="cuda"):
                logits = model(img_eff, img_vit)
            logits_f = torch.clamp(logits.float(), min=-50, max=50)
            step_probs.extend(F.softmax(logits_f, dim=1)[:, 1].cpu().numpy())
            step_labels.extend(labels.numpy())

        step_probs = np.array(step_probs)
        if all_probs_accum is None:
            all_probs_accum = step_probs
            all_labels      = np.array(step_labels)
        else:
            all_probs_accum += step_probs

    avg_probs  = all_probs_accum / n_steps
    y_pred_tta = (avg_probs >= 0.5).astype(int)
    return (accuracy_score(all_labels, y_pred_tta),
            f1_score(all_labels, y_pred_tta, average="macro"),
            roc_auc_score(all_labels, avg_probs),
            all_labels, y_pred_tta, avg_probs)


# ─────────────────────────────────────────
# CELL 14 — Training Loop
# ─────────────────────────────────────────
best_val_f1   = -1.0
wait          = 0
best_path     = os.path.join(SAVE_DIR, "fusion_image-best.pt")
best_f1_epoch = 1
train_accs, val_accs = [], []

print("=" * 60)
print("Starting Fusion Image Model Training")
print("=" * 60)

for epoch in range(1, EPOCHS + 1):

    fusion_model.train()
    # Keep backbones in eval mode always — they are frozen
    fusion_model.effnet.eval()
    fusion_model.vit.eval()

    running_loss = 0.0
    correct = total = 0

    pbar = tqdm(train_loader, desc=f"Epoch {epoch:02d}/{EPOCHS}")
    for img_eff, img_vit, labels in pbar:
        img_eff = img_eff.to(device)
        img_vit = img_vit.to(device)
        labels  = labels.to(device)

        img_eff, img_vit, labels_a, labels_b, lam = mixup_data(
            img_eff, img_vit, labels)

        optimizer.zero_grad(set_to_none=True)

        with autocast(device_type="cuda"):
            logits = fusion_model(img_eff, img_vit)
            loss   = mixup_criterion(criterion, logits, labels_a, labels_b, lam)

        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(fusion_model.parameters(), GRAD_CLIP)
        scaler.step(optimizer)
        scaler.update()

        running_loss += loss.item()
        with torch.no_grad():
            preds = torch.argmax(logits.float(), dim=1)
        correct += (lam * (preds == labels_a).float() +
                    (1 - lam) * (preds == labels_b).float()).sum().item()
        total   += labels_a.size(0)

        pbar.set_postfix({
            "loss": f"{running_loss / (pbar.n or 1):.4f}",
            "acc" : f"{correct / total:.4f}"
        })

    train_acc = correct / total
    train_accs.append(train_acc)

    val_loss, val_f1, val_acc, val_auc, _, _, _ = evaluate(fusion_model, val_loader)
    val_accs.append(val_acc)

    scheduler.step(val_f1)

    print(f"\nEpoch {epoch:02d} | "
          f"train_acc={train_acc:.4f} | "
          f"val_acc={val_acc:.4f} | "
          f"val_F1={val_f1:.4f} | "
          f"val_AUC={val_auc:.4f} | "
          f"val_loss={val_loss:.4f}")

    if val_f1 > best_val_f1:
        best_val_f1   = val_f1
        best_f1_epoch = epoch
        wait = 0
        torch.save({
            "epoch"           : epoch,
            "model_state_dict": fusion_model.state_dict(),
            "val_f1"          : val_f1,
            "val_acc"         : val_acc,
            "feature_dim"     : fusion_model.feature_dim,
        }, best_path)
        print(f"✅ Best model saved  →  val_F1={val_f1:.4f}")
    else:
        wait += 1
        print(f"⏳ No improvement ({wait}/{PATIENCE})")
        if wait >= PATIENCE:
            print("⏹️  Early stopping triggered.")
            break

print(f"\nBest Val macro-F1: {best_val_f1:.4f}  (epoch {best_f1_epoch})")


# ─────────────────────────────────────────
# CELL 15 — Training vs Validation Accuracy Plot
# ─────────────────────────────────────────
best_epoch = best_f1_epoch

plt.figure(figsize=(9, 5))
plt.plot(range(1, len(train_accs)+1), [a*100 for a in train_accs],
         label="Training Accuracy", linewidth=2)
plt.plot(range(1, len(val_accs)+1),   [a*100 for a in val_accs],
         label="Validation Accuracy", linewidth=2, color="orange")
plt.axvline(x=best_epoch, linestyle="--", color="gray")
plt.scatter([best_epoch], [val_accs[best_epoch-1]*100],
            color="red", zorder=5, s=80,
            label=f"Best Saved (Epoch {best_epoch})\n"
                  f"Val: {val_accs[best_epoch-1]*100:.2f}%\n"
                  f"Train: {train_accs[best_epoch-1]*100:.2f}%")
plt.xlabel("Epoch"); plt.ylabel("Accuracy (%)")
plt.title("Fusion Image Model — Training vs Validation Accuracy")
plt.legend(); plt.grid(alpha=0.3)
plt.tight_layout()
acc_plot_path = os.path.join(SAVE_DIR, "fusion_image_train_val_accuracy.png")
plt.savefig(acc_plot_path, dpi=400, bbox_inches="tight")
plt.show()
print("Saved →", acc_plot_path)


# ─────────────────────────────────────────
# CELL 16 — Load Best & Evaluate on Test
# ─────────────────────────────────────────
checkpoint = torch.load(best_path, map_location=device, weights_only=True)
fusion_model.load_state_dict(checkpoint["model_state_dict"])
fusion_model.to(device)
print(f"Loaded best model from epoch {checkpoint['epoch']}")

test_loss, test_f1, test_acc, test_auc, y_true, y_pred, y_prob = evaluate(
    fusion_model, test_loader)

print(f"\n── Standard Inference ──────────────────────────────────")
print(f"TEST  →  Accuracy: {test_acc:.4f}  |  Macro-F1: {test_f1:.4f}  |  ROC-AUC: {test_auc:.4f}")
print("\nClassification Report:\n",
      classification_report(y_true, y_pred,
                             target_names=["FALSE", "TRUE"], digits=4))
print("Confusion Matrix:\n", confusion_matrix(y_true, y_pred))

print(f"\n── Test-Time Augmentation ({TTA_STEPS} passes) ──────────────────")
tta_acc, tta_f1, tta_auc, tta_true, tta_pred, tta_prob = evaluate_tta(
    fusion_model, test_df, IMG_DIR, n_steps=TTA_STEPS)
print(f"TTA   →  Accuracy: {tta_acc:.4f}  |  Macro-F1: {tta_f1:.4f}  |  ROC-AUC: {tta_auc:.4f}")
print("\nTTA Classification Report:\n",
      classification_report(tta_true, tta_pred,
                             target_names=["FALSE", "TRUE"], digits=4))
print("TTA Confusion Matrix:\n", confusion_matrix(tta_true, tta_pred))


# ─────────────────────────────────────────
# CELL 17 — Train vs Test ROC Curve
# ─────────────────────────────────────────
@torch.no_grad()
def get_roc_data(loader, name="", mdl=None):
    if mdl is None: mdl = fusion_model
    mdl.eval()
    y_true_list, y_score_list = [], []
    for img_eff, img_vit, labels in tqdm(loader, desc=f"ROC ({name})"):
        img_eff = img_eff.to(device)
        img_vit = img_vit.to(device)
        with autocast(device_type="cuda"):
            logits = mdl(img_eff, img_vit)
        probs = F.softmax(torch.clamp(logits.float(), min=-50, max=50),
                          dim=1)[:, 1].cpu().numpy()
        y_true_list.extend(labels.numpy())
        y_score_list.extend(probs)
    fpr, tpr, _ = roc_curve(y_true_list, y_score_list)
    return fpr, tpr, auc(fpr, tpr)

fpr_train, tpr_train, auc_train = get_roc_data(train_loader, "Train")
fpr_test,  tpr_test,  auc_test  = get_roc_data(test_loader,  "Test")

plt.figure(figsize=(7, 6))
plt.plot(fpr_train, tpr_train, label=f"Train (AUC = {auc_train:.4f})", linewidth=2.5)
plt.plot(fpr_test,  tpr_test,  label=f"Test  (AUC = {auc_test:.4f})",  linewidth=2.5)
plt.plot([0,1], [0,1], linestyle="--", color="gray")
plt.xlabel("False Positive Rate"); plt.ylabel("True Positive Rate")
plt.title("Train vs Test ROC Curve — Fusion Image Model")
plt.legend(loc="lower right"); plt.grid(alpha=0.3)
plt.tight_layout()
roc_path = os.path.join(SAVE_DIR, "fusion_image_train_vs_test_roc.png")
plt.savefig(roc_path, dpi=400, bbox_inches="tight")
plt.show()
print("Saved →", roc_path)

gap = abs(auc_train - auc_test)
print(f"\nTrain AUC : {auc_train:.4f}")
print(f"Test  AUC : {auc_test:.4f}")
print(f"Gap       : {gap:.4f}  →  ", end="")
if   gap <= 0.02: print("Excellent generalization ✅")
elif gap <= 0.05: print("Slight overfitting — acceptable ⚠️")
else:             print("Significant overfitting — review regularization 🚨")


# ─────────────────────────────────────────
# CELL 18 — Save Per-Sample Predictions
# ─────────────────────────────────────────
preds_df = pd.DataFrame({
    "y_true"      : y_true,
    "y_prob_TRUE" : y_prob,
    "y_pred"      : y_pred,
    "y_pred_tta"  : tta_pred,
    "y_prob_tta"  : tta_prob,
})
preds_path = os.path.join(SAVE_DIR, "fusion_image_test_predictions.csv")
preds_df.to_csv(preds_path, index=False)
print("✅ Saved test predictions →", preds_path)


# ─────────────────────────────────────────
# CELL 19 — Final Summary
# ─────────────────────────────────────────
print("\n" + "="*60)
print("     Fusion Image Model — Final Results Summary")
print("="*60)
print(f"  Individual Models")
print(f"    EfficientNet-B3  Val F1 : {eff_ckpt['val_f1']:.4f}")
print(f"    ViT-Base-384     Val F1 : {vit_ckpt['val_f1']:.4f}")
print(f"  Fusion Image Model")
print(f"    Test Accuracy   :  {test_acc:.4f}")
print(f"    Test Macro-F1   :  {test_f1:.4f}")
print(f"    Test ROC-AUC    :  {test_auc:.4f}")
print(f"    Train ROC-AUC   :  {auc_train:.4f}")
print(f"  TTA ({TTA_STEPS} passes)")
print(f"    TTA Accuracy    :  {tta_acc:.4f}")
print(f"    TTA Macro-F1    :  {tta_f1:.4f}")
print(f"    TTA ROC-AUC     :  {tta_auc:.4f}")
print(f"  Best Val F1       :  {best_val_f1:.4f}  (epoch {best_f1_epoch})")
print(f"  Model saved at    :  {best_path}")
print(f"  Feature dim       :  {fusion_model.feature_dim}  (for Final Multimodal)")
print("="*60)
```
